````python
import numpy as np

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

from xgboost import XGBClassifier
````

````python
data = load_breast_cancer()

x = data.data
y = data.target

print(x.shape)
print(y.shape)
````

````python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
````

````python
model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)
````

````python
model.fit(x_train, y_train)
````

````python
y_train_pred = model.predict(x_train)
y_pred = model.predict(x_test)
````

````python
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_pred)

print(f"Train Accuracy: {train_accuracy:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
````

````python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
````

````python
learning_rates = [0.01, 0.05, 0.1, 0.2]
n_estimators_values = [50, 100, 200, 500]

for learning_rate in learning_rates:
    for n_estimators in n_estimators_values:

        model = XGBClassifier(
            learning_rate=learning_rate,
            n_estimators=n_estimators,
            max_depth=3,
            random_state=42
        )

        model.fit(x_train, y_train)

        y_train_pred = model.predict(x_train)
        y_pred = model.predict(x_test)

        train_accuracy = accuracy_score(y_train, y_train_pred)
        test_accuracy = accuracy_score(y_test, y_pred)

        print(
            f"learning_rate={learning_rate:<4} | "
            f"n_estimators={n_estimators:<3} | "
            f"Train={train_accuracy:.4f} | "
            f"Test={test_accuracy:.4f}"
        )
````

````python
depth_values = [1, 2, 3, 4, 5, 7, 10]

for depth in depth_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=depth,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"max_depth={depth:<2} | "
        f"Train={train_accuracy:.4f} | "
        f"Test={test_accuracy:.4f}"
    )
````

````python
subsample_values = [0.5, 0.6, 0.7, 0.8, 0.9, 1.0]

for subsample in subsample_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        subsample=subsample,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"subsample={subsample:<3} | "
        f"Train={train_accuracy:.4f} | "
        f"Test={test_accuracy:.4f}"
    )
````

````python
colsample_values = [0.5, 0.6, 0.7, 0.8, 0.9, 1.0]

for colsample in colsample_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=colsample,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"colsample_bytree={colsample:<3} | "
        f"Train={train_accuracy:.4f} | "
        f"Test={test_accuracy:.4f}"
    )
````

````python
alpha_values = [0, 0.01, 0.1, 0.5, 1, 5, 10]

for alpha in alpha_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=1.0,
        reg_alpha=alpha,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"reg_alpha={alpha:<5} | "
        f"Train={train_accuracy:.4f} | "
        f"Test={test_accuracy:.4f}"
    )
````

````python
lambda_values = [0, 0.1, 1, 5, 10, 50, 100]

for reg_lambda in lambda_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=1.0,
        reg_lambda=reg_lambda,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"reg_lambda={reg_lambda:<5} | "
        f"Train={train_accuracy:.4f} | "
        f"Test={test_accuracy:.4f}"
    )
````

````python
gamma_values = [0, 0.1, 0.5, 1, 5, 10]

for gamma in gamma_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=1.0,
        gamma=gamma,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"gamma={gamma:<5} | "
        f"Train={train_accuracy:.4f} | "
        f"Test={test_accuracy:.4f}"
    )
````

# XGBoost — Notes

## 1. What is XGBoost?

**XGBoost (Extreme Gradient Boosting)** is an optimized implementation of the **Gradient Boosting** algorithm.

It is a **supervised learning** algorithm used for:

* Classification
* Regression

XGBoost builds multiple decision trees **sequentially**, where each new tree attempts to correct the errors made by the previous trees.

---

## 2. Basic Idea

Instead of building one large decision tree:

```text
Dataset
   ↓
Decision Tree
   ↓
Prediction
```

XGBoost builds many trees:

```text
Dataset
   ↓
Tree 1
   ↓
Tree 2 → corrects errors from Tree 1
   ↓
Tree 3 → corrects remaining errors
   ↓
...
   ↓
Final Prediction
```

Each tree contributes to the final prediction.

---

# 3. XGBoost vs Gradient Boosting

Both are based on the same fundamental idea:

> Build trees sequentially to correct previous errors.

XGBoost adds several improvements such as:

* Regularization
* Efficient computation
* Parallel processing
* Row/feature sampling
* Powerful hyperparameter control

# 5. Important XGBoost Parameters

## `n_estimators`

Number of trees built by the model.

```python
XGBClassifier(n_estimators=100)
```

Higher value:

```text
More trees
   ↓
Potentially better learning
   ↓
But increased risk of overfitting
```

---

## `learning_rate`

Controls how much each tree contributes to the final model.

```python
XGBClassifier(learning_rate=0.1)
```

Small learning rate:

```text
Small contribution per tree
        ↓
Usually need more trees
```

Large learning rate:

```text
Larger contribution per tree
        ↓
Fewer trees may be required
        ↓
Can overfit more easily
```

`learning_rate` and `n_estimators` usually need to be considered together.

---

## `max_depth`

Controls the maximum depth of each decision tree.

```python
XGBClassifier(max_depth=3)
```

Higher `max_depth`:

```text
More complex trees
       ↓
Can learn complex patterns
       ↓
Higher overfitting risk
```

Lower `max_depth`:

```text
Simpler trees
       ↓
Less complexity
       ↓
Can underfit if too small
```

---

# 6. `subsample`

Controls the fraction of **training samples (rows)** used for each tree.

```python
XGBClassifier(subsample=0.8)
```

For example:

```text
subsample = 1.0
→ 100% of rows

subsample = 0.8
→ 80% of rows
```

Using fewer rows introduces randomness and can help reduce overfitting.

---

# 7. `colsample_bytree`

Controls the fraction of **features (columns)** used by each tree.

```python
XGBClassifier(colsample_bytree=0.8)
```

For example:

```text
colsample_bytree = 1.0
→ 100% of features

colsample_bytree = 0.7
→ 70% of features
```

It introduces feature-level randomness and can improve generalization.

### Remember

```text
subsample
→ Rows

colsample_bytree
→ Columns
```

---

# 8. `reg_alpha` — L1 Regularization

`reg_alpha` controls the strength of **L1 regularization**.

```python
XGBClassifier(reg_alpha=1)
```

Connection:

```text
Lasso Regression
      ↓
L1 Regularization
      ↓
XGBoost reg_alpha
```

Increasing `reg_alpha` makes the model more strongly penalized for complexity.

Too much regularization can cause **underfitting**.

---

# 9. `reg_lambda` — L2 Regularization

`reg_lambda` controls the strength of **L2 regularization**.

```python
XGBClassifier(reg_lambda=10)
```

Connection:

```text
Ridge Regression
      ↓
L2 Regularization
      ↓
XGBoost reg_lambda
```

Increasing it can reduce model complexity.

Too much can lead to underfitting.

---

# 10. `gamma`

`gamma` controls the **minimum loss reduction required to make a new tree split**.

```python
XGBClassifier(gamma=1)
```

Conceptually:

```text
gamma ↑
   ↓
Harder to create a split
   ↓
Less tree complexity
```

Therefore:

* Low `gamma` → splits are easier
* High `gamma` → splits require more improvement

Too high → possible underfitting.

---

# 11. XGBoost Regularization Summary

| Parameter    | Purpose                                  |
| ------------ | ---------------------------------------- |
| `reg_alpha`  | L1 regularization                        |
| `reg_lambda` | L2 regularization                        |
| `gamma`      | Controls whether a split is worth making |

---

# 12. Hyperparameter Tuning

Instead of manually trying parameters, we can use:

```python
GridSearchCV
```

or:

```python
RandomizedSearchCV
```

---

# 13. GridSearchCV

GridSearchCV tries **every combination** of the specified parameter values.

Example:

```python
param_grid = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.05, 0.1],
    'max_depth': [2, 3],
    'subsample': [0.8, 1.0]
}
```

Number of combinations:

```text
3 × 2 × 2 × 2 = 24
```

With:

```python
cv=5
```

each combination is evaluated using 5-fold cross-validation:

```text
24 combinations × 5 folds
= 120 model fits
```

We don't manually run them.

```python
grid_search.fit(x_train, y_train)
```

automatically performs them.

---

# 14. RandomizedSearchCV

RandomizedSearchCV does **not necessarily test every combination**.

Instead, we specify how many combinations to try:

```python
n_iter=10
```

For example:

```text
10 combinations × 5 folds
= 50 model fits
```

So:

```text
GridSearchCV
→ Tests every combination

RandomizedSearchCV
→ Tests a selected number of combinations
```

### Comparison

|                       | GridSearchCV | RandomizedSearchCV |
| --------------------- | ------------ | ------------------ |
| Search                | Exhaustive   | Random             |
| Computation           | Higher       | Usually lower      |
| Speed                 | Slower       | Faster             |
| Large parameter space | Expensive    | More practical     |

---

# 15. Cross-Validation

Cross-validation evaluates the model multiple times using different portions of the **training data**.

For `cv=5`:

```text
x_train
   ↓
┌────────┬────────┬────────┬────────┬────────┐
│ Fold 1 │ Fold 2 │ Fold 3 │ Fold 4 │ Fold 5 │
└────────┴────────┴────────┴────────┴────────┘
```

Each round uses:

```text
4 folds → Training
1 fold  → Validation
```

The validation fold changes every round.

**Important:**

```text
x_train → Cross-validation
x_test  → Final evaluation
```

The test set should remain untouched during hyperparameter selection.

---

# 16. `best_params_`

After GridSearchCV or RandomizedSearchCV:

```python
print(grid_search.best_params_)
```

Example:

```text
    'n_estimators': 200,
    'subsample': 0.8
}
```

---

# 17. `best_score_`

```python

```

Returns the **best average cross-validation score**.

Example:

```text

```

```text

```

---

# 18. `best_estimator_`

```python
best_model = grid_search.best_estimator_
```

Returns the model configured with the best parameters found by the search.

We can then use:

```python
y_pred = best_model.predict(x_test)
```
