# Linear Regression — Complete Notes

## 1. What is Linear Regression?

**Linear Regression** is a supervised machine learning algorithm used to predict a **continuous numerical target**.

Examples:

* House price prediction
* Salary prediction
* Sales prediction
* Temperature prediction
* Medical outcome prediction

The model learns a relationship between **input features (X)** and a **continuous target (y)**.

### Core idea

Given:

```text
Features ──────► Linear Regression ──────► Predicted Target
```

The model learns the best mathematical relationship between inputs and output from training data.

---

# 2. Simple Linear Regression

With one feature:

$$
\hat{y} = b_0 + b_1x
$$

Where:

* \(\hat y\) = predicted output
* \(b_0\) = intercept
* \(b_1\) = coefficient/slope
* \(x\) = input feature

### Example

$$
\hat y = 25 + 10x
$$

If:

$$
x=6
$$

then:

$$
\hat y=25+10(6)=85
$$

### Intercept

The intercept is the predicted value of \(y\) when **all input features are 0**.

It is **not necessarily the minimum value** of the target.

### Coefficient

The coefficient tells us how much the **predicted target changes for a one-unit increase in that feature**, assuming other features remain constant in multiple regression.

---

# 3. Multiple Linear Regression

Real-world problems usually have multiple features.

$$
\hat y=b_0+b_1x_1+b_2x_2+\cdots+b_nx_n
$$

Example:

$$
Salary=20000+3000(Experience)+5000(SkillScore)
$$

If:

* Experience = 4
* SkillScore = 6

then:

$$
Salary=20000+3000(4)+5000(6)
$$

$$
=62000
$$

### Coefficient interpretation

The ₹5,000 coefficient means:

> For a one-unit increase in SkillScore, predicted salary increases by ₹5,000, **holding Experience constant**.

It does **not** mean that a person with SkillScore = 1 earns ₹5,000.

---

# 4. Important: Association ≠ Causation

A Linear Regression coefficient tells us about a relationship in the model.

It does **not automatically prove causation**.

For example:

> House size has a positive coefficient.

This does not prove:

> Increasing house size will definitely cause the house price to increase.

Other variables, confounding factors, data collection, and model assumptions matter.

---

# 5. What does Linear Regression Learn?

During:

```python
model.fit(X_train, y_train)
```

the model learns:

* coefficients/weights
* intercept

You can inspect them with:

```python
model.coef_
model.intercept_
```

Prediction:

```python
y_pred = model.predict(X_test)
```

---

# 6. Residuals

A **residual** is the difference between actual and predicted value.

$$
e_i=y_i-\hat y_i
$$

Where:

* \(y_i\) = actual value
* \(\hat y_i\) = predicted value
* \(e_i\) = residual

### Example

Actual = 100
Predicted = 90

$$
e=100-90=10
$$

Positive residual → model **underpredicted**.

Actual = 100
Predicted = 110

$$
e=100-110=-10
$$

Negative residual → model **overpredicted**.

---

# 7. Ordinary Least Squares — OLS

Linear Regression commonly uses **Ordinary Least Squares (OLS)**.

The objective is to find coefficients and intercept that minimize the **sum of squared residuals**.

$$
SSE=\sum_{i=1}^{n}(y_i-\hat y_i)^2
$$

Therefore:

$$
\min \sum_{i=1}^{n}(y_i-\hat y_i)^2
$$

### Why square the errors?

Because positive and negative residuals can cancel.

Example:

```text
+10 + (-10) = 0
```

That would incorrectly suggest no error.

Squaring gives:

```text
10² + (-10)² = 200
```

Large errors are also penalized more heavily.

---

# 8. Regression Evaluation Metrics

## MAE — Mean Absolute Error

$$
MAE=\frac{1}{n}\sum|y_i-\hat y_i|
$$

It represents the **average absolute prediction error**.

Example:

```text
Actual:     100   200   300
Predicted:   90   220   270

Errors:      10   -20    30
Absolute:    10    20    30
```

$$
MAE=\frac{10+20+30}{3}=20
$$

### Interpretation

> On average, predictions are off by about 20 target units in absolute terms.

It does **not** mean every prediction is exactly 20 units wrong.

---

# 9. MSE — Mean Squared Error

$$
MSE=\frac{1}{n}\sum(y_i-\hat y_i)^2
$$

MSE:

* squares errors
* prevents cancellation
* penalizes large errors more strongly

Lower MSE is generally better when comparing models on the same target/test set.

---

# 10. RMSE — Root Mean Squared Error

$$
RMSE=\sqrt{MSE}
$$

RMSE is useful because it is expressed in the **same units as the target**.

Example:

```text
RMSE = 52
```

means the model's prediction errors have a typical magnitude around that scale, though RMSE is **not** the exact error of every prediction.

### MAE vs RMSE

| Metric | Main idea                            |
| ------ | ------------------------------------ |
| MAE    | Average absolute error               |
| MSE    | Average squared error                |
| RMSE   | Square root of average squared error |

**MAE** treats errors more linearly.

**MSE/RMSE** give larger errors more influence.

---

# 11. R² — Coefficient of Determination

$$
R^2=1-\frac{SS_{res}}{SS_{tot}}
$$

R² compares the model against a baseline that simply predicts the **mean target value**.

### Interpretation

If:

$$
R^2=0.85
$$

we say:

> The model explains approximately 85% of the variance in the target relative to the mean-prediction baseline.

⚠️ R² is **not accuracy**.

R² can also be negative on test data when the model performs worse than the mean baseline in squared-error terms.

---

# 12. MAE vs RMSE vs R²

Think of them as answering different questions:

```text
MAE / RMSE
     ↓
"How large are my prediction errors?"

R²
     ↓
"How much target variation does my model explain
relative to the mean baseline?"
```

Do not automatically choose a model based on only one metric.

---

# 13. Linear Regression Assumptions

## 1. Linearity 🔴

The relationship between features and target should be reasonably approximated by a linear relationship.

It does **not** mean every data point must lie perfectly on a straight line.

A strongly curved relationship may indicate that ordinary Linear Regression is not sufficient.

---

## 2. Independence 🔴

Observations/errors should generally be independent.

This can become problematic with:

* time-series data
* repeated measurements
* related observations

---

## 3. Homoscedasticity 🔴

Residual variance should be roughly constant.

Good:

```text
     •  •     •
  •     •  •
---------------- 0
    •   •    •
```

Potential problem:

```text
 •
   •
     •
       •
         •
```

A funnel-shaped spread suggests **heteroscedasticity** — the error variance changes as predictions change.

---

## 4. Multicollinearity 🔴

Multicollinearity occurs when predictor features are strongly correlated with each other.

Example:

```text
Experience ↔ Years in Industry
correlation = 0.94
```

This can cause:

* unstable coefficients
* difficult coefficient interpretation
* coefficients sensitive to small changes in data

Important:

> Multicollinearity does not necessarily mean the model's predictions will be terrible.

It primarily creates problems with **individual coefficient estimates and interpretation**.

### Quick screening

```python
X.corr()
```

This checks **feature-to-feature correlation**.

```python
df.corr()
```

checks correlations among all numerical columns.

Correlation is only a screening method. **VIF** can provide a more comprehensive assessment.

---

## 5. Residual Behavior 🔴

Residuals should ideally look approximately random.

### Random scatter

Generally a good sign.

### Clear pattern

May indicate the model is missing some structure.

For example:

```text
Curved residual pattern
        ↓
Possible nonlinear relationship
```

### Key idea

> **Random residuals → model has captured the structure reasonably well.**

> **Patterned residuals → model may be missing some structure.**

---

## 6. Normality of Residuals 🟡

Residual normality is mainly important for **statistical inference**.

For pure predictive ML, perfectly normal residuals are not a strict requirement.

---

# 14. Diagnostic Plots


Ideal predictions lie close to:

$$

### Question it answers:

## Residual Plot

```python

```

### Question it answers:

> **Are my errors behaving randomly and sensibly?**

Look for:

| Pattern                   | Possible meaning                             |
| ------------------------- | -------------------------------------------- |
| Random scatter            | Good sign                                    |
| Curved pattern            | Possible non-linearity                       |
| Funnel                    | Possible heteroscedasticity                  |
| Clusters                  | Missing structure/groups                     |
| Strong systematic pattern | Model may be missing important relationships |

# 15. Generalization

**Generalization** means performing well on **unseen data**.

The goal of ML is not simply:

> "Get the lowest training error."

The goal is:

> **Learn patterns that work on new data.**

---

# 16. Overfitting

Overfitting occurs when the model learns the training data too specifically, including noise or accidental patterns.

Typical pattern:

```text
Training performance → Very high
Test performance     → Much lower
```

Example:

```text
Model A:
Train R² = 0.92
Test R²  = 0.90

Model B:
Train R² = 0.99
Test R²  = 0.55
```

Model B shows much stronger signs of overfitting.

Model A is more trustworthy because its train and test performance are close and both are good.

---

# 17. Underfitting

Underfitting occurs when the model is too simple to capture the underlying patterns.

Typical pattern:

```text
Training performance → Poor
Test performance     → Poor
```

### Goal

```text
Good training performance
          +
Good unseen/test performance
          ↓
     Good Generalization
```

---

# 18. Feature Scaling

Basic `LinearRegression` does **not generally require feature scaling** for the standard OLS solution.

However, scaling becomes important in many other algorithms and techniques, such as:

* Ridge
* Lasso
* Gradient-based optimization
* KNN
* SVM

So:

> Don't memorize "Linear Regression always needs scaling."

# 22. Important Mistakes to Avoid

### ❌ Mistake 1: Intercept = minimum value

Wrong.

The intercept is the predicted target when all input features are zero.

---

### ❌ Mistake 2: R² = accuracy

Wrong.

$$
R^2=0.90
$$

does not mean 90% accuracy.

---

### ❌ Mistake 3: Raw residuals can be averaged

Positive and negative errors can cancel.

Use:

* absolute errors → MAE
* squared errors → MSE/RMSE

---

### ❌ Mistake 4: Higher R² automatically means better model

Not necessarily.

A model with higher R² can have worse prediction error depending on the situation.

Use multiple appropriate metrics and consider the real-world objective.

---

### ❌ Mistake 5: Large coefficient = most important feature

Not necessarily.

Coefficient magnitude depends on:

* feature units
* feature scale
* relationships between predictors

Raw coefficient magnitudes should not automatically be used to rank feature importance.

---

### ❌ Mistake 6: Correlation proves causation

It doesn't.

Regression coefficients describe modeled associations, not automatically causal relationships.

---

### ❌ Mistake 7: Patterned residuals = model is simply "bad"

Not necessarily.

A residual pattern tells you that the model may be **missing structure**.

For example, a curved pattern can suggest non-linearity.

---

### ❌ Mistake 8: 2D `y` is forbidden

It isn't.

For single-target regression, the conventional form is:

```python
y = df["target"]
```

giving:

```text
(n_samples,)
```

while:

```python
y = df[["target"]]
```

gives:

```text
(n_samples, 1)
```

Both can often work, but 1D `y` is the standard convention for single-target regression.

# 23. What You Need to Know for Interviews

### What is Linear Regression?

> Linear Regression is a supervised learning algorithm used to predict continuous numerical targets by modeling the relationship between input features and the target using a linear function.

### What does OLS do?

> Ordinary Least Squares estimates the coefficients and intercept by minimizing the sum of squared residuals.

### What is a residual?

> A residual is the difference between the actual value and the predicted value.

$$
e=y-\hat y
$$

### MAE vs RMSE?

> MAE measures average absolute error, while RMSE is based on squared errors and therefore gives larger errors more influence.

### What does R² mean?

> R² measures how much target variance the model explains relative to a baseline that always predicts the mean target value.

### What does a residual pattern indicate?

> A systematic residual pattern suggests that the model may be missing some structure, such as non-linearity or changing error variance.

### What is multicollinearity?

> Multicollinearity occurs when predictor variables are strongly correlated, making individual coefficients unstable and difficult to interpret.
