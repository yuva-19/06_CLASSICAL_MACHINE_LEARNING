# 1️⃣ Linear Regression

### What?

Used to predict a **continuous numerical value**.

Example:

```text
House features → House price
Experience → Salary
Area → Rent
```

### Main idea

Find a line that minimizes prediction error.

```text
y = b₀ + b₁x
```

For multiple features:

```text
y = b₀ + b₁x₁ + b₂x₂ + ... + bₙxₙ
```

### Important concept

Linear Regression minimizes:

**MSE (Mean Squared Error)**

### Strengths

* Simple
* Fast
* Easy to interpret
* Good baseline model

### Weakness

* Assumes a linear relationship
* Sensitive to outliers
* Can suffer from multicollinearity

---

# 2️⃣ Ridge Regression

### What?

Linear Regression + **L2 regularization**.

```text
Linear Regression
       +
L2 penalty
       ↓
Ridge Regression
```

The penalty discourages very large coefficients.

### Parameter

```python
Ridge(alpha=1)
```

`alpha` controls regularization strength.

```text
alpha ↑
   ↓
Stronger regularization
   ↓
Coefficients shrink
```

### Important

Ridge generally **doesn't make coefficients exactly zero**.

### Use when

* Features are correlated
* Linear Regression is overfitting
* You want to control large coefficients

---

# 3️⃣ Logistic Regression

Despite its name, Logistic Regression is mainly used for **classification**.

Example:

```text
Email → Spam / Not Spam
Customer → Buy / Not Buy
Tumor → Benign / Malignant
```

It uses the **sigmoid function**:

```text
          1
σ(z) = ────────
       1 + e⁻ᶻ
```

Output:

```text
0 → 1 probability
```

Then a threshold is used to determine the class.

For example:

```text
Probability > 0.5 → Class 1
Probability < 0.5 → Class 0
```

### Important metrics

* Accuracy
* Precision
* Recall
* F1-score
* Confusion Matrix

---

# 4️⃣ SVM — Support Vector Machine

### Main idea

Find the **best decision boundary** separating classes.

```text
Class A       |       Class B
              |
              |
         Decision Boundary
```

SVM tries to maximize the **margin** between the classes.

### Important concepts

**Support vectors**

The data points closest to the decision boundary.

**Margin**

Distance between the decision boundary and the closest points.

### Kernel

Kernels allow SVM to handle nonlinear relationships.

Common kernels:

```text
linear
poly
rbf
sigmoid
```

### Important parameters

```text
C
gamma
kernel
```

`C` controls the trade-off between:

```text
larger margin
vs
classification errors
```

---

# 5️⃣ Decision Tree

A Decision Tree makes predictions using a sequence of questions.

Example:

```text
Age > 30?
   ↓
 Yes        No
 ↓           ↓
Income?    Class A
```

It repeatedly splits the data.

### Classification

Common splitting concepts:

* Gini impurity
* Entropy
* Information Gain

### Regression

Uses measures such as MSE to determine useful splits.

### Important parameters

```text
max_depth
min_samples_split
min_samples_leaf
```

### Major weakness

Decision Trees can **overfit easily**.

```text
Very deep tree
     ↓
Learns training data extremely well
     ↓
Poor generalization
```

---

# 6️⃣ Random Forest

Random Forest = **many Decision Trees + randomness + aggregation**

```text
             Dataset
                ↓
       ┌────────┼────────┐
       ↓        ↓        ↓
     Tree 1   Tree 2   Tree 3
       ↓        ↓        ↓
       └────────┼────────┘
                ↓
          Final Prediction
```

### Why "Random"?

Two important sources of randomness:

1. Random samples of rows
2. Random subsets of features

### Classification

Trees vote:

```text
Tree 1 → A
Tree 2 → B
Tree 3 → A
Tree 4 → A
       ↓
Final → A
```

### Regression

Predictions are generally averaged.

### Advantages

* Reduces overfitting compared with a single tree
* Handles nonlinear relationships
* Works well with many features
* Provides feature importance

---

# 7️⃣ Gradient Boosting

Gradient Boosting builds trees **sequentially**.

```text
Tree 1
  ↓
Errors
  ↓
Tree 2 learns from errors
  ↓
Errors
  ↓
Tree 3 learns from remaining errors
  ↓
Final model
```

Unlike Random Forest:

```text
Random Forest
→ Trees mainly work independently

Gradient Boosting
→ Trees are built sequentially
```

### Important parameters

```text
n_estimators
learning_rate
max_depth
```

### Important concept

Small learning rate usually requires more trees.

```text
learning_rate ↓
      ↓
n_estimators often ↑
```

---

# 8️⃣ Lasso Regression

Lasso = Linear Regression + **L1 regularization**

```text
Linear Regression
       +
L1 penalty
       ↓
Lasso
```

### Important property

Lasso can make coefficients **exactly zero**.

Example:

```text
Feature 1 → 4.5
Feature 2 → 0
Feature 3 → 2.1
Feature 4 → 0
```

So Lasso can perform a form of **feature selection**.

### Parameter

```python
Lasso(alpha=1)
```

Higher `alpha`:

```text
Stronger regularization
       ↓
More coefficients can become zero
```

Too much → underfitting.

---

# 9️⃣ Elastic Net

Elastic Net combines:

**L1 + L2 regularization**

```text
        Elastic Net
             │
       ┌─────┴─────┐
       ↓           ↓
      L1           L2
    Lasso        Ridge
```

### Main parameters

```text
alpha
l1_ratio
```

`alpha`:

```text
Overall regularization strength
```

`l1_ratio`:

```text
How much L1 vs L2
```

Conceptually:

```text
l1_ratio = 1
→ Lasso-like

l1_ratio = 0
→ Ridge-like
```

### Why use it?

Useful when you have:

* Many features
* Correlated features
* Need for regularization
* Some desire for feature selection

# 🔟 KNN — K-Nearest Neighbors

KNN predicts based on the **nearest data points**.

Example:

```text
New point
   ↓
Find nearest K points
   ↓
Look at their classes
   ↓
Majority vote
```

If:

```text
K = 5
```

the model looks at the 5 nearest neighbors.

### Important parameter

```text
n_neighbors
```

### Small K

```text
K ↓
   ↓
Very sensitive to individual points
   ↓
Can overfit
```

### Large K

```text
K ↑
   ↓
More generalized
   ↓
Can underfit
```

### Very important

KNN is **distance-based**, so feature scaling is usually important.

Common distance:

```text
Euclidean distance
```

KNN can be used for:

```text
Classification
Regression
```

# 1️⃣1️⃣ XGBoost

XGBoost = **Extreme Gradient Boosting**

It is an optimized and highly regularized implementation of gradient boosting.

```text
Tree 1
  ↓
Tree 2 corrects errors
  ↓
Tree 3 corrects errors
  ↓
...
  ↓
Final prediction
```

### Important parameters

| Parameter          | Meaning                                  |
| ------------------ | ---------------------------------------- |
| `n_estimators`     | Number of trees                          |
| `learning_rate`    | Contribution of each tree                |
| `max_depth`        | Maximum tree depth                       |
| `subsample`        | Fraction of rows used                    |
| `colsample_bytree` | Fraction of features used                |
| `reg_alpha`        | L1 regularization                        |
| `reg_lambda`       | L2 regularization                        |
| `gamma`            | Minimum improvement required for a split |

### Important connections

```text
Lasso
  ↓
L1
  ↓
XGBoost reg_alpha
```

```text
Ridge
  ↓
L2
  ↓
XGBoost reg_lambda
```

# 🔥 GridSearchCV & RandomizedSearchCV

We also used XGBoost to understand hyperparameter tuning.

### GridSearchCV

Tests **every combination** we provide.

```text
24 combinations
×
5 folds
=
120 model fits
```

We don't manually run 120 times.

```python
grid_search.fit(x_train, y_train)
```

does it automatically.

### RandomizedSearchCV

Tests a specified number of randomly selected combinations.

For example:

```text
n_iter = 10
cv = 5

10 × 5
= 50 fits
```

### Difference

```text
GridSearchCV
→ Exhaustive
→ More expensive

RandomizedSearchCV
→ Random subset
→ Usually faster
```

```text
Need feature selection?
        ↓
Lasso
```

```text
Need L1 + L2?
        ↓
Elastic Net
```

```text
Distance/neighborhood relationship?
        ↓
KNN
```

```text
Want simple nonlinear model?
        ↓
Decision Tree
```

```text
Want robust tree ensemble?
        ↓
Random Forest
```

```text
Want sequential boosting?
        ↓
Gradient Boosting
```

# 🚨 Most Important Differences

### Ridge vs Lasso

```text
Ridge → L2 → shrinks coefficients
Lasso → L1 → can make coefficients 0
```

### Random Forest vs Gradient Boosting

```text
Random Forest
→ Many trees
→ More independent
→ Bagging

Gradient Boosting
→ Trees sequentially
→ Each tree improves previous errors
→ Boosting
```

### Gradient Boosting vs XGBoost

```text
Gradient Boosting
→ General boosting approach

XGBoost
→ Optimized / regularized gradient boosting implementation
```

### KNN vs Tree

```text
KNN
→ Distance-based
→ Scaling important

Tree
→ Rule/split-based
→ Scaling usually not necessary
```

### Linear vs Tree models

```text
Linear
→ Learns linear relationships

Tree
→ Can learn nonlinear relationships
```
