````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
````

````python
iris = load_iris()

x = iris.data
y = iris.target
````

````python
print(x.shape)
print(y.shape)
````

````python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
````

````python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])
````

````python
pipeline.fit(x_train, y_train)
````

````python
y_train_pred = pipeline.predict(x_train)
y_pred = pipeline.predict(x_test)
````

````python
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_pred)

print(f"Train Accuracy: {train_accuracy:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
````

````python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
````

````python
k_values = [1, 3, 5, 7, 9, 11, 15, 21]

for k in k_values:

    pipeline = Pipeline([
        ("scaler", StandardScaler()),
        ("knn", KNeighborsClassifier(n_neighbors=k))
    ])

    pipeline.fit(x_train, y_train)

    y_train_pred = pipeline.predict(x_train)
    y_pred = pipeline.predict(x_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_pred)

    print(
        f"K={k:<3} | "
        f"Train Accuracy={train_accuracy:.4f} | "
        f"Test Accuracy={test_accuracy:.4f}"
    )
````

# K-Nearest Neighbors (KNN)

## 1. Definition

**K-Nearest Neighbors (KNN)** is a supervised learning algorithm that makes predictions based on the **closest training samples**.

Core idea:

> **Similar data points tend to have similar outputs.**

KNN is a **distance-based** and **instance-based (lazy learning)** algorithm.

Unlike models such as Linear Regression or Decision Trees, KNN does not learn an explicit equation or tree structure during training. It primarily stores the training data and uses it during prediction.

---

## 2. How KNN Works

For a new data point:

```text
New data point
      ↓
Calculate distance to training points
      ↓
Find K nearest neighbors
      ↓
Make prediction
```

The prediction depends on the problem:

```text
Classification → Majority vote
Regression     → Average target values
```

---

## 3. What is K?

`K` represents the **number of nearest neighbors** considered.

Example:

```python
KNeighborsClassifier(n_neighbors=5)
```

means the model considers the **5 closest training samples**.

### Classification Example

```text
K = 5

A
A
B
A
B

A = 3
B = 2

Prediction → A
```

KNN classification uses **majority voting**.

### Regression Example

```text
K = 3

100
120
110

Prediction = (100 + 120 + 110) / 3
           = 110
```

KNN regression generally uses the **average target value** of the neighbors.

---

# 4. Distance Metrics

KNN determines "nearest" using a **distance metric**.

## Euclidean Distance ⭐

The most commonly used distance metric.

For two points:

$$
A=(x_1,y_1)
$$

$$
B=(x_2,y_2)
$$

Euclidean distance:

$$
d(A,B)=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

It represents the **straight-line distance** between two points.

---

## Manhattan Distance

Manhattan distance measures the total movement along each dimension.

$$
d(A,B)=|x_2-x_1|+|y_2-y_1|
$$

Example:

```python
KNeighborsClassifier(
    n_neighbors=5,
    metric="manhattan"
)
```

Common metrics to understand:

```text
Euclidean
Manhattan
```

---

# 5. Feature Scaling ⭐⭐⭐

KNN is a **distance-based algorithm**, so feature scaling is extremely important.

Suppose:

```text
Age     → 20–80
Salary  → 20,000–200,000
```

Salary has a much larger numerical range and can dominate the distance calculation.

Therefore, we generally scale the features before KNN.

```text
Original features
       ↓
StandardScaler
       ↓
KNN
```

A Pipeline is preferred:

```python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])
```

This also helps prevent **data leakage**, because the scaler is fitted using the training data within the pipeline.

# 8. Effect of K

Choosing K creates a **bias-variance trade-off**.

## Small K

Example:

```text
K = 1
```

The model is highly sensitive to individual training samples.

Characteristics:

* More flexible
* Lower bias
* Higher variance
* Higher risk of overfitting

```text
Small K
   ↓
High variance
   ↓
Overfitting risk
```

## Large K

Example:

```text
K = 50
```

The model considers many neighbors.

Characteristics:

* Smoother predictions
* Higher bias
* Lower variance
* Too large → underfitting

```text
Large K
   ↓
High bias
   ↓
Underfitting risk
```

### Important

Don't say:

> K=1 always overfits.

Instead:

> **Small K has a higher risk of overfitting because it is highly sensitive to individual training samples.**

---

# 9. Choosing K

We should not simply assume a particular K is best.

Try different values:

```python
k_values = [1, 3, 5, 7, 9, 11, 15, 21]
```

Then compare model performance.

A proper workflow is:

```text
Training Data
      ↓
Cross-Validation
      ↓
Try different K values
      ↓
Choose suitable K
      ↓
Final Test Evaluation
```

The test set should ideally be kept untouched until final evaluation.

---

# 10. Important KNN Hyperparameters

## `n_neighbors`

Controls K.

```python
n_neighbors=5
```

> How many neighbors should be considered?

---

## `weights`

Controls the influence of neighbors.

### `weights="uniform"`

All neighbors have equal influence.

```python
weights="uniform"
```

### `weights="distance"`

Closer neighbors have greater influence.

```python
weights="distance"
```

Easy memory:

```text
uniform  → Equal influence
distance → Closer = more influence
```

---

## `metric`

Controls how distance is calculated.

```python
metric="euclidean"
```

or:

```python
metric="manhattan"
```

Easy memory:

```text
n_neighbors → How many?
weights      → How much influence?
metric       → How is distance calculated?
```

# 14. Computational Cost

KNN can become expensive during prediction because it needs to calculate distances to training samples to identify the nearest neighbors.

As the dataset grows:

```text
Training samples ↑
       ↓
Distance calculations ↑
       ↓
Prediction cost ↑
```

Therefore, KNN may not be ideal for extremely large datasets.

---

# 15. Curse of Dimensionality

KNN can also struggle with **high-dimensional data**.

As the number of features increases:

```text
Features ↑
   ↓
Data becomes sparse in feature space
   ↓
Distance becomes less meaningful
   ↓
KNN performance can decrease
```

Therefore, dimensionality reduction or feature selection may sometimes be useful.

---

# 16. Advantages

* Simple and intuitive
* Easy to implement
* Can model nonlinear relationships
* Works for classification and regression
* Doesn't require a strong assumption of linearity
* Useful for relatively small/medium datasets

---

# 17. Limitations

* Prediction can be computationally expensive
* Sensitive to feature scaling
* Sensitive to irrelevant features
* Choice of K affects performance
* Can suffer from the curse of dimensionality
* Can perform poorly with very large datasets
