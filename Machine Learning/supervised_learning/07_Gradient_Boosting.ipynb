````python
import pandas as pd

df = pd.read_csv(r"Y:\Python\AI ML\Month 2\week 6-Classic ML\Machine Learning\supervised_learning\data\raw\bank-full.csv", sep =';')
df.head()
````

````python
x = df.drop('y',axis=1)
y = df['y']
````

````python
categorical_columns = x.select_dtypes(include="object").columns
numerical_columns = x.select_dtypes(exclude="object").columns

print("Categorical:", list(categorical_columns))
print("Numerical:", list(numerical_columns))
````

````python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import GradientBoostingClassifier

preprocessor = ColumnTransformer(
    transformers=[
        ('cat',OneHotEncoder(handle_unknown='ignore'),categorical_columns)
    ]
)

pipeline = Pipeline([
    ('preprocessor',preprocessor),
    ('model',GradientBoostingClassifier(
        n_estimators= 100,
        learning_rate= 0.1,
        max_depth= 3,
        random_state= 42
    ))
])
````

````python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    random_state= 42,
    stratify= y
)
````

````python
pipeline.fit(x_train,y_train)
````

````python
y_train_pred = pipeline.predict(x_train)
y_pred = pipeline.predict(x_test)
````

````python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print('Accuracy Score: ')
training_accuracy = accuracy_score(y_train,y_train_pred)
test_accuracy = accuracy_score(y_test,y_pred)
print(f'Train Accuracy: {training_accuracy}')
print(f'Test Accuracy: {test_accuracy}')
print()

print('Confusion Matrix: ')
print(confusion_matrix(y_test,y_pred))
print()
print('Classification Report: ')
print(classification_report(y_test,y_pred))
````

````python
from sklearn.metrics import balanced_accuracy_score

balanced_acc = balanced_accuracy_score(y_test, y_pred)

print(f'Balanced Accuracy: {balanced_acc}')
````

````python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import GradientBoostingClassifier

configs = [
    (0.1, 50),
    (0.1, 100),
    (0.1, 200),
    (0.05, 200),
    (0.01, 500)
]

for lr, n in configs:

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", GradientBoostingClassifier(
            n_estimators=n,
            learning_rate=lr,
            max_depth=3,
            random_state=42
        ))
    ])

    model.fit(x_train, y_train)

    train_acc = model.score(x_train, y_train)
    test_acc = model.score(x_test, y_test)

    print(
        f"learning_rate={lr} | "
        f"n_estimators={n}| "
        f"Train={train_acc:.4f} | "
        f"Test={test_acc:.4f}"
    )
````

# Gradient Boosting — Notes

## 1. What is Gradient Boosting?

**Gradient Boosting** is an **ensemble learning algorithm** that combines multiple weak decision trees to create a strong predictive model.

Unlike Random Forest, where trees are largely built independently, Gradient Boosting builds trees **sequentially**.

### Core idea

> Each new tree tries to improve/correct the mistakes made by the existing ensemble.

```text
Tree 1
  ↓
Current prediction
  ↓
Calculate errors / loss
  ↓
Tree 2 learns correction
  ↓
Updated prediction
  ↓
Tree 3 learns remaining correction
  ↓
...
```

---

# 2. Boosting

**Boosting** combines multiple weak learners sequentially.

The important idea is:

```text
Previous model → New model improves it → New model improves it again → ...
```

This differs from **Bagging**.

### Bagging — Random Forest

```text
Tree 1 ─┐
Tree 2 ─┤
Tree 3 ─┼──→ Aggregate predictions
Tree 4 ─┤
Tree 5 ─┘
```

### Boosting — Gradient Boosting

```text
Tree 1 → Tree 2 → Tree 3 → Tree 4 → ...
          ↑
     improves previous model
```

---

# 3. How Gradient Boosting Works

At a high level:

### Step 1 — Start with an initial model

The model makes initial predictions.

### Step 2 — Calculate the loss

Measure how wrong the current model is using a **loss function**.

### Step 3 — Calculate the gradient

The gradient indicates the **direction in which the loss should be changed to improve the model**.

### Step 4 — Train a new decision tree

The new tree learns to reduce the existing model's errors.

### Step 5 — Add the new tree to the existing model

The new tree's contribution is controlled by the **learning rate**.

### Step 6 — Repeat

This process continues for the specified number of boosting stages.

---

# 4. Sequential Learning

Gradient Boosting is **sequential**.

```text
Model 1
   ↓
Model 2
   ↓
Model 3
   ↓
Model 4
   ↓
Final Model
```

Each stage builds on the previous stages.

This is why Gradient Boosting is generally **harder to parallelize during training** than Random Forest.

---

# 5. Weak Learners

Gradient Boosting commonly uses **decision trees as weak learners**.

The trees are usually relatively shallow.

Example:

```python
GradientBoostingClassifier(
    max_depth=3
)
```

Shallow trees help control model complexity and reduce overfitting.

---

# 6. Important Hyperparameters

## `n_estimators`

Number of boosting stages/trees.

```python
n_estimators=100
```

Means approximately **100 sequential trees** are used.

Increasing `n_estimators` can improve learning, but too many trees can lead to overfitting.

---

## `learning_rate`

Controls how much each new tree contributes to the final model.

```python
learning_rate=0.1
```

### High learning rate

```text
Each tree → larger contribution
```

Can learn faster but may overfit.

### Low learning rate

```text
Each tree → smaller contribution
```

Usually requires more trees.

### Important relationship

> **Lower learning rate generally requires more estimators.**

Example:

```text
learning_rate = 0.1
n_estimators = 100
```

versus

```text
learning_rate = 0.05
n_estimators = 200
```

---

## `max_depth`

Controls the depth of the individual decision trees.

```python
max_depth=3
```

Higher depth:

* More complex trees
* Can learn more detailed patterns
* Greater risk of overfitting

Lower depth:

* Simpler trees
* Less complexity
* May underfit if too small

---

## `subsample`

Controls the fraction of training samples used for each boosting stage.

Example:

```python
subsample=0.7
```

Approximately **70% of the training samples** are used at each stage.

Using subsampling can help reduce overfitting and introduce additional randomness.

---

# 7. Main Hyperparameter Relationship

A very important concept:

```text
Learning Rate ↓
       ↓
Need more trees
       ↓
More gradual learning
```

For example:

```text
learning_rate = 0.1
n_estimators = 100
```

can be comparable in learning capacity to a configuration with a smaller learning rate and more trees, although the results are not guaranteed to be identical.

---

# 8. Gradient Boosting vs Random Forest

| Feature             | Random Forest                  | Gradient Boosting                             |
| ------------------- | ------------------------------ | --------------------------------------------- |
| Ensemble method     | Bagging                        | Boosting                                      |
| Trees               | Largely independent            | Sequential                                    |
| Main idea           | Aggregate many trees           | Correct/improve previous model                |
| Training            | Parallelizable                 | Sequential                                    |
| `n_estimators`      | Number of trees                | Number of boosting stages                     |
| `learning_rate`     | No                             | Yes                                           |
| Feature randomness  | Important                      | Not the same mechanism as RF                  |
| Main benefit        | Reduces variance               | Sequentially improves predictions             |
| Overfitting control | Depth, features, samples, etc. | Depth, learning rate, estimators, subsampling |

### Easy memory trick

**Random Forest:**

> "Many trees give their opinions."

**Gradient Boosting:**

> "Each new tree fixes/improves what came before."

---

# 9. AdaBoost vs Gradient Boosting

Both are **boosting algorithms**, but their mechanisms differ.

### AdaBoost

Focuses more on **incorrectly classified samples** by adjusting their importance/weights.

### Gradient Boosting

Uses the **gradient of the loss function** to determine how the model should improve.

| AdaBoost                         | Gradient Boosting         |
| -------------------------------- | ------------------------- |
| Focuses on misclassified samples | Optimizes a loss function |
| Adjusts sample importance        | Uses loss gradients       |
| Sequential                       | Sequential                |

# 11. Using Pipeline

For datasets containing categorical features, preprocessing and the model can be combined using a pipeline.

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    ))
])
```

Train:

```python
pipeline.fit(X_train, y_train)
```

Predict:

```python
y_pred = pipeline.predict(X_test)
```

### Important

`train_test_split()` normally happens **before** the pipeline:

```text
Raw Data
   ↓
train_test_split()
   ↓
X_train / X_test
   ↓
Pipeline
   ↓
Preprocessing
   ↓
Gradient Boosting
```

---

# 12. Why Use a Pipeline?

A pipeline keeps preprocessing and model training together.

```text
X_train
   ↓
Fit preprocessing
   ↓
Transform
   ↓
Train model
```

For test data:

```text
X_test
   ↓
Use already-fitted preprocessing
   ↓
Transform
   ↓
Predict
```

This helps prevent **data leakage** and ensures consistent preprocessing.

# 15. Hyperparameter Experiment

We tested:

```text
learning_rate=0.1 | n_estimators=50
learning_rate=0.1 | n_estimators=100
learning_rate=0.1 | n_estimators=200
learning_rate=0.05 | n_estimators=200
learning_rate=0.01 | n_estimators=500
```

Results showed that increasing the number of trees **does not automatically improve test performance**.

For example:

```text
0.1 + 100 trees → Test ≈ 0.8949
0.1 + 200 trees → Test ≈ 0.8944
```

Training performance increased while test performance slightly decreased, showing the beginning of overfitting.

# 16. Advantages

* Strong predictive performance
* Can model complex nonlinear relationships
* Works well with tabular data
* Sequentially improves weak learners
* Can capture interactions between features
* Provides feature importance

---

# 17. Disadvantages

* Training is sequential
* Generally slower than highly parallelizable Random Forest
* Sensitive to hyperparameter choices
* Can overfit with excessive complexity
* Requires careful tuning of learning rate, number of trees, and tree depth

---

# 18. Key Concepts to Remember

### Gradient Boosting

> **Sequential ensemble of decision trees where each new tree improves the existing model by minimizing the loss.**

### Bagging

> **Build largely independent models and aggregate their predictions.**

### Boosting

> **Build models sequentially so later models improve the existing ensemble.**

### Learning Rate

> Controls the contribution of each new tree.

### `n_estimators`

> Number of boosting stages/trees.

### `max_depth`

> Controls the complexity of each tree.

### `subsample`

> Fraction of training samples used at each boosting stage.

### Most important distinction

```text
Random Forest
= Bagging
= Independent trees
= Aggregate

Gradient Boosting
= Boosting
= Sequential trees
= Correct/improve
```
