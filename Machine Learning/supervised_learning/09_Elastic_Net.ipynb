````python
import numpy as np
import pandas as pd

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error, r2_score
````

````python
x, y = make_regression(
    n_samples=500,
    n_features=10,
    n_informative=5,
    noise=20,
    random_state=42
)
````

````python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
````

````python
elastic = ElasticNet(
    alpha=1,
    l1_ratio=0.5
)
````

````python
elastic.fit(x_train, y_train)
````

````python
y_train_pred = elastic.predict(x_train)
y_pred = elastic.predict(x_test)
````

````python
train_r2 = r2_score(y_train, y_train_pred)
test_r2 = r2_score(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

print(f"Train R2: {train_r2:.4f}")
print(f"Test R2: {test_r2:.4f}")
print(f"MSE: {mse:.4f}")
````

````python
print(elastic.coef_)
````

````python
zero_coefficients = np.sum(elastic.coef_ == 0)

print("Zero coefficients:", zero_coefficients)
````

````python
l1_ratios = [0, 0.25, 0.5, 0.75, 1]

for ratio in l1_ratios:

    model = ElasticNet(
        alpha=1,
        l1_ratio=ratio
    )

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_pred)

    zero_coefficients = np.sum(model.coef_ == 0)

    print(
        f"l1_ratio={ratio:<4} | "
        f"Zero coefficients={zero_coefficients:<3} | "
        f"Train R2={train_r2:.4f} | "
        f"Test R2={test_r2:.4f}"
    )
````

````python
alphas = [0.01, 0.1, 1, 10]
l1_ratios = [0.25, 0.5, 0.75, 1]

for alpha in alphas:
    for ratio in l1_ratios:

        model = ElasticNet(
            alpha=alpha,
            l1_ratio=ratio
        )

        model.fit(x_train, y_train)

        y_train_pred = model.predict(x_train)
        y_pred = model.predict(x_test)

        train_r2 = r2_score(y_train, y_train_pred)
        test_r2 = r2_score(y_test, y_pred)

        zero_coefficients = np.sum(model.coef_ == 0)

        print(
            f"alpha={alpha:<5} | "
            f"l1_ratio={ratio:<4} | "
            f"zeros={zero_coefficients:<3} | "
            f"Train R2={train_r2:.4f} | "
            f"Test R2={test_r2:.4f}"
        )
````

````python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import r2_score

models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(alpha=1),
    "Lasso": Lasso(alpha=1),
    "Elastic Net": ElasticNet(alpha=0.1, l1_ratio=0.75)
}

for name, model in models.items():

    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_pred = model.predict(x_test)

    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_pred)

    print(
        f"{name:<20} | "
        f"Train R2={train_r2:.4f} | "
        f"Test R2={test_r2:.4f}"
    )
````

# Elastic Net Regression

## 1. Definition

**Elastic Net Regression** is a regularized version of Linear Regression that combines **L1 and L2 regularization**.

It combines the main benefits of:

* **Lasso** → L1 regularization and feature selection
* **Ridge** → L2 regularization and coefficient shrinkage

> **Elastic Net = L1 + L2 Regularization**

---

## 2. Why Elastic Net?

Lasso and Ridge each have different strengths.

### Lasso

* Performs feature selection.
* Can make coefficients exactly `0`.
* Can sometimes be too aggressive when features are correlated.

### Ridge

* Shrinks coefficients.
* Handles correlated features well.
* Generally keeps all features.

Elastic Net combines both approaches.

```text
             Elastic Net
                  │
          ┌───────┴───────┐
          ↓               ↓
         L1              L2
          ↓               ↓
       Lasso            Ridge
          ↓               ↓
 Feature selection   Coefficient
                     shrinkage
```

---

## 3. Elastic Net Objective Function

$$
Loss =
MSE +
\alpha
\left[
l1\_ratio\sum|w_i|
+
(1-l1\_ratio)\sum w_i^2
\right]
$$

The important idea is that Elastic Net contains **both L1 and L2 penalties**.

---

## 4. `alpha`

`alpha` controls the **overall strength of regularization**.

### Small `alpha`

* Weak regularization
* Model behaves more like Linear Regression
* Coefficients are affected less

### Large `alpha`

* Strong regularization
* Coefficients shrink more
* More risk of underfitting

```text
alpha ↑
  ↓
Regularization ↑
  ↓
Coefficients shrink more
  ↓
Too large → Underfitting
```

---

## 5. `l1_ratio` ⭐

`l1_ratio` controls the **balance between L1 and L2 regularization**.

### `l1_ratio = 1`

```text
100% L1 + 0% L2
       ↓
Lasso
```

### `l1_ratio = 0`

```text
0% L1 + 100% L2
       ↓
Ridge
```

### `l1_ratio = 0.5`

```text
50% L1 + 50% L2
       ↓
Elastic Net
```

Examples:

| `l1_ratio` |   L1 |   L2 |
| ---------: | ---: | ---: |
|          0 |   0% | 100% |
|       0.25 |  25% |  75% |
|        0.5 |  50% |  50% |
|       0.75 |  75% |  25% |
|          1 | 100% |   0% |

### Easy memory trick

> **`alpha` = strength**
> **`l1_ratio` = mixture**

---

## 6. Feature Selection

Because Elastic Net contains an **L1 component**, it can make some coefficients exactly zero.

Example:

```text
Feature 1 → 12.4
Feature 2 → 0
Feature 3 → -5.2
Feature 4 → 0
```

Features 2 and 4 are effectively excluded from the model.

Therefore:

> Elastic Net can perform **automatic feature selection**.

---

---

## 10. Linear vs Ridge vs Lasso vs Elastic Net

| Model             | Regularization | Feature Selection |
| ----------------- | -------------- | ----------------- |
| Linear Regression | None           | ❌                 |
| Ridge             | L2             | ❌                 |
| Lasso             | L1             | ✅                 |
| Elastic Net       | L1 + L2        | ✅                 |

### Core difference

```text
Linear Regression
        ↓
No regularization

Ridge
        ↓
L2
        ↓
Shrink coefficients

Lasso
        ↓
L1
        ↓
Shrink + select features

Elastic Net
        ↓
L1 + L2
        ↓
Shrink + select features
```

---

## 11. When to Use Each

### Linear Regression

Use when:

* You want a simple baseline.
* Regularization isn't necessary.

### Ridge

Use when:

* Features are correlated.
* You want coefficient shrinkage.
* You generally want to keep all features.

### Lasso

Use when:

* You want automatic feature selection.
* You want a sparse model.

### Elastic Net

Use when:

* You want feature selection **and** coefficient stabilization.
* You have correlated features.
* Pure Lasso may be too aggressive.

Elastic Net had the highest test R² in our particular experiment, but the differences were very small.

---

## 13. Key Takeaways

* **Elastic Net = L1 + L2 regularization**
* L1 comes from **Lasso**
* L2 comes from **Ridge**
* `alpha` controls **overall regularization strength**
* `l1_ratio` controls **L1 vs L2 balance**
* L1 can make coefficients exactly `0`
* Therefore Elastic Net can perform **feature selection**
* Large `alpha` can cause **underfitting**
* `l1_ratio = 0` → Ridge-like
* `l1_ratio = 1` → Lasso
* Cross-validation can be used to find good `alpha` and `l1_ratio`
* Feature scaling is generally important when using regularized linear models.
